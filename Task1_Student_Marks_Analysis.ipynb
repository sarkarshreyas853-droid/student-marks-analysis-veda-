# Task 1: Student Marks Analysis

This notebook analyzes student marks and calculates the **total**, **average**, **highest**, and **lowest** marks.

## Step 1: Store the marks

We store the marks of each student in a Python list. The data is manually created.

```python
marks = [78, 92, 65, 88, 74, 95, 81, 59, 70, 86]
print("Marks:", marks)
```

## Step 2: Calculate total, average, highest and lowest

- `sum()` gives the total.
- The average is the total divided by the number of students.
- `max()` gives the highest mark and `min()` gives the lowest.

```python
total = sum(marks)
average = total / len(marks)
highest = max(marks)
lowest = min(marks)

print("Number of students:", len(marks))
print("Total marks:", total)
print("Average marks:", round(average, 2))
print("Highest marks:", highest)
print("Lowest marks:", lowest)
```

## Step 3: Show the results in a table-like format

```python
print(f"{'Result':<12}{'Value':>10}")
print('-' * 22)
print(f"{'Total':<12}{total:>10}")
print(f"{'Average':<12}{average:>10.2f}")
print(f"{'Highest':<12}{highest:>10}")
print(f"{'Lowest':<12}{lowest:>10}")
```

## Interview questions

**How do you calculate the average of a list in Python?**
Use `sum(list) / len(list)`.

**What is the difference between `max()` and `min()`?**
`max()` returns the largest item in a list. `min()` returns the smallest item.

**How can Python be used to analyze student data?**
Store the data in lists or dictionaries, then use built-in functions or libraries such as pandas to calculate totals, averages, and rankings.
